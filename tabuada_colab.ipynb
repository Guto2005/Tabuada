# 📱 Compilar APK — Tabuada BGMax

## Passo 1 — Faça upload dos seus arquivos
Antes de rodar qualquer célula, clique no ícone de **pasta** à esquerda,
depois em **Upload** e envie estes arquivos:
- `main.py`
- `fundo-inicial.png`
- `fundo-tabuada.png`
- `bg.png`
- `bgspl.png`
- `buildozer.spec`

Depois rode as células abaixo **em ordem**.

```
# Passo 2 — Instalar dependências do sistema
%%bash
sudo apt-get update -qq
sudo apt-get install -y \
    python3-pip build-essential git python3 python3-dev \
    ffmpeg libsdl2-dev libsdl2-image-dev libsdl2-mixer-dev \
    libsdl2-ttf-dev libportmidi-dev libswscale-dev libavformat-dev \
    libavcodec-dev zlib1g-dev openjdk-17-jdk autoconf libtool \
    pkg-config libncurses5-dev libncursesw5-dev libtinfo5 cmake \
    libffi-dev libssl-dev zip unzip > /dev/null 2>&1
echo '✅ Dependências do sistema instaladas'
```

```
# Passo 3 — Instalar Buildozer e Cython
%%bash
pip install -q buildozer cython==0.29.33
echo '✅ Buildozer instalado'
```

```
# Passo 4 — Copiar arquivos do projeto para a pasta de build
import os, shutil

# Pasta do projeto
os.makedirs('/content/tabuada', exist_ok=True)

# Arquivos que você fez upload ficam em /content/
arquivos = [
    'main.py',
    'fundo-inicial.png',
    'fundo-tabuada.png',
    'bg.png',
    'bgspl.png',
    'buildozer.spec'
]

for arq in arquivos:
    origem = f'/content/{arq}'
    destino = f'/content/tabuada/{arq}'
    if os.path.exists(origem):
        shutil.copy(origem, destino)
        print(f' {arq} copiado')
    else:
        print(f' {arq} NÃO encontrado — verifique o upload!')
```

```
# Passo 5 — Compilar o APK (isso demora 15-30 minutos na primeira vez)
%%bash
cd /content/tabuada
buildozer -v android debug 2>&1 | tail -50
```

```
# Passo 6 — Baixar o APK gerado
import glob
from google.colab import files

apks = glob.glob('/content/tabuada/bin/*.apk')

if apks:
    print(f'✅ APK encontrado: {apks[0]}')
    files.download(apks[0])
else:
    print('❌ APK não gerado. Role para cima e veja o erro na célula anterior.')
```

## ❓ Deu erro?

Os erros mais comuns e como resolver:

| Erro | Solução |
|------|---------|
| `SDK license not accepted` | Já está resolvido no buildozer.spec |
| `Could not find kivymd` | Normal, o buildozer baixa automaticamente |
| `No module named gtts` | Já está no requirements do buildozer.spec |
| Sessão expirou | Rode novamente — o Colab tem limite de tempo |
| Erro de NDK | Aguarde, o download é grande (~1GB) |

**Na segunda compilação é muito mais rápido** porque o cache fica salvo.
